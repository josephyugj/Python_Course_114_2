### 程式實例 ch5_4.py: 奇數偶數的判斷，下列5-8行是傳統用法，第10-13是符合PEP8用法，第15行事Python高手用法。

```python
# ch5_4.py
print("奇數偶數判斷")
num = input("請輸入任意整值: ")
rem = int(num) % 2
if (rem == 0):
    print("%d 是偶數" % int(num))
else:
    print("%d 是奇數" % int(num))
# PEP 8
if rem:
    print("%d 是奇數" % int(num))
else:
    print("%d 是偶數" % int(num))
# 高手用法
print("%d 是奇數" % int(num) if rem else "%d 是偶數" % int(num))
```

### 程式實例 ch5_6.py: 有一地區的票價收費標準是100元。
- 但是如果小於等於6歲或大於等於80歲，收費是打2哲。

## 5.6 巢狀的if敘述
### 程式實例 ch5_8.py: 測試每一年是否閏年，閏年的條件是首先可以被4整除(相當於沒有餘數)，這個條件成立時，還必須符合，它除以100時餘數不是0或是除以400時餘數為0，當2個條件皆符合時才算閏年。

```python
# ch5_8.py
print("判斷輸入年份是否潤年")
year = input("請輸入年分: ")
rem4 = int(year) % 4
rem100 = int(year) % 100
rem400 = int(year) % 400
if rem4 == 0:
    if rem100 != 0 or rem400 == 0:
        print("%s 是潤年" % year)
    else:
        print("%s 不是潤年" % year)
else:
    print("%s 不是潤年" % year)
```

```python
# ch5_10.py
ans = 0                         # 讀者心中的數字
print("猜數字遊戲,請心中想一個 0 - 7之間的數字, 然後回答問題")

truefalse = "輸入y或Y代表有, 其它代表無 : "
# 檢測2進位的第1位是否含1
q1 = "有沒有看到心中的數字 : \n" + \
     "1, 3, 5, 7 \n"
num = input(q1 + truefalse)
print(num)
if num == "y" or num == "Y":
    ans += 1
# 檢測2進位的第2位是否含1
truefalse = "輸入y或Y代表有, 其它代表無 : "
q2 = "有沒有看到心中的數字 : \n" + \
     "2, 3, 6, 7 \n"
num = input(q2 + truefalse)
if num == "y" or num == "Y":
    ans += 2
# 檢測2進位的第3位是否含1
truefalse = "輸入y或Y代表有, 其它代表無 : "
q3 = "有沒有看到心中的數字 : \n" + \
     "4, 5, 6, 7 \n"
num = input(q3 + truefalse)
if num == "y" or num == "Y":
    ans += 4

print("讀者心中所想的數字是 : ", ans)
```

### 5.8.4 求一元二次方程式的根
### 程式實例 ch5_13.py: 有一個一元二次方程式如下，求這個方程式的根
$ 3x^2 + 5x + 1 = 0$

```python
# ch5_13.py
a = 3
b = 5
c = 1

r1 = (-b + (b**2-4*a*c)**0.5)/(2*a)
r2 = (-b - (b**2-4*a*c)**0.5)/(2*a)
print("r1 = %6.4f,    r2 = %6.4f" % (r1, r2))
```

### 5.8.5 求解聯立線性方程式 
### 程式實例 ch5_14.py: 假設有一個聯立線性方程式如下:
$ ax + by = e$
$ cx + dy = f$

可以用下列方程式獲得 x 和 y 值

$ x = \frac{e*d+b*f}{a*d-b*c} $

$ y = \frac{a*f-e*c}{a*d-b*c} $ 

在公式中，如果 "a*d-b*c" 等於，則此聯立線性方程式無解。

```python
# ch5_14.py
a = 2
b = 3
c = 1
d = -2
e = 13
f = -4

x = (e*d - b*f) / (a*d - b*c)
y = (a*f - e*c) / (a*d - b*c)
print("x = %6.4f,    y = %6.4f" % (x, y))
```
